### 🚩작업형 제 2유형

제공된 학습용 데이터(10_2_train.csv)는 여러 상권 내 특선(상권 유형, 건물 면적, 건물 연식, 세대 수 등)과 연간 총 가스 소비량 정보를 담고 있다. 학습용 데이터를 활용하여 건물의 연간 총 가스 소비량(gas_totl)을 예측하는 모델을 개발하고, 이 중 가장 우수한 모델을 평가용 데이터(10_2_test.csv)에 적용하여 예측 결과를 제출하시오.   
% 평가지표 : RMSE

Data description
- biz_type : 상권 유형
- area : 건물 면적
- age : 건물 연식
- num_households : 건물 내 세대수
- gas_totl : 건물의 연간 총 가스 소비량 (결측치는 0으로 대체되어 있음.)

제출형식
- CSV 파일명 : result.csv(디렉토리/폴더명 불포함)
- 예측 칼럼명 : pred
- 제출 칼럼 개수 : pred 1개
- 평가용 데이터 개수와 예측 결과 데이터 개수 일치

```python
import pandas as pd
import numpy as np

train = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/10_2_train.csv')
test = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/10_2_test.csv')

train.shape
```

```python
# @title
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import OneHotEncoder, StandardScaler
# ❗해당 패키지 위치에 make_column_transformer 있음!
from sklearn.compose import make_column_transformer

X_train = train.drop('gas_totl', axis = 1)
y_train = train['gas_totl']

X_test = test.drop('gas_totl', axis = 1)

cal_cols = X_train.select_dtypes('object').columns
num_cols = ['area', 'num_households']

# 결측치 없는 것 확인.
# print(X_train.isnull().sum())
# print(X_test.isnull().sum())

transformer = make_column_transformer(
    (OneHotEncoder(handle_unknown = 'ignore', sparse_output = False), cal_cols),
    (StandardScaler(), num_cols)
).set_output(transform = 'pandas') # 판다스 변환 굳이 할 필요 없긴 함.

X_train_pre = transformer.fit_transform(X_train)
X_test_pre = transformer.transform(X_test)

rf_model = RandomForestRegressor()
rf_model.fit(X_train_pre, y_train)

y_pred = rf_model.predict(X_test_pre)

pd.DataFrame({'pred': y_pred}).to_csv('result.csv', index = False)
```

 #### 🚩 1. 한 기업이 인사관리 데이터를 가지고 이직 여부 예측 모델을 개발하려고 한다. 다음 문제를 풀이하시오.

데이터 : 10_3_1.csv

Data description
- attrition : 이직여부(0=잔류, 1=이직)
- age : 나이
- income : 연봉
- overtime : 야근 상태(0=해당없음, 1=보통, 2=상시 등)

1-1. 이직 여부를 예측하는 로지스틱 회귀 모형을 적합하고, 유의한 변수(유의확률 0.05 미만)의 회귀계수를 소수점 셋째 자리까지 반올림하여 제출하시오. (단, 절편 제외)

1-2. age가 1 증가할 때 이직(또는 잔류) 오즈비(odds ratio)를 소수점 셋째 자리까지 반올림하여 제출하시오.

1-3. age = 20, income = 3000, overtime = 2 값을 가진 데이터의 이직확률을 모델로 예측하여 소수점 셋째 자리까지 반올림하여 제출하시오.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/10_3_1.csv')

df.head()
```

```python
# @title
import numpy as np
from statsmodels.formula.api import logit

# ❗범주형 변수는 사용할 때 C()를 해줘야 함!
model = logit('attrition ~ age + income + C(overtime)', data = df).fit()

# 유의한 변수. income
#1-1 정답 : -0.005
print(round(model.params['income'], 3))

# #1-2 정답 : 0.894
print(round(np.exp(model.params['age']), 3))

predict_var = pd.DataFrame({'age' : [20], 'income' : [3000], 'overtime' : [2]})
pred_y = model.predict(predict_var)

# #1-3 정답 : 0.48
print(round(pred_y[0], 3))

# model.summary()
```

- height : 집 높이

2-1. 주택 가격을 예측하는 다중선형회귀모형을 적합하고, 유의한 변수(유의확률 0.05 미만)의 회귀계수 합(절편 제외)을 소수점 셋째 자리까지 반올림하여 제출하시오.

2-2. 유의한 변수만으로 다중선형회귀모형을 다시 적합하고, 결정계수를 소수점 셋째 자리까지 반올림하여 제출하시오.

2-3. area=100, height=10, wall=1 값을 가진 데이터의 예측 주택 가격을 모델로 예측하여 소수점 셋째 자리까지 반올림하여 제출하시오.(단, 이전 문제에서 뽑은 통계적으로 유의미한 변수만 선택할 것)

```python
import pandas as pd
import numpy as np

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/10_3_2.csv')

df.head()
```

```python
# @title
from statsmodels.formula.api import ols

model = ols('price ~ area + height + wall', data = df).fit()
# 유의한 변수 : area height
# 2-1 정답 : 10.289
print(round(model.params[['area', 'height']].sum(), 3))

model2 = ols('price ~ area + height', data = df).fit()
# 2-2 정답 : 0.859
print(round(model2.rsquared, 3))

predict_var = pd.DataFrame({'area' : [100], 'height' : [10]})
y_pred = model2.predict(predict_var)

# 2-3 정답 : 329.036
print(round(y_pred[0], 3))
```

#### 🚩3. 제시된 문제를 순서대로 풀고, 해답을 제시하시오.(9_1_3)

3-1. 결측치 처리

- 근속연수 : 결측치는 각 부서와 등급별 평균 근속연수로 채우시오.(평균값의 소수점은 버림 처리)

3-2. 조건에 따른 평균 계산
- A : 부서가 'HR'이고 등급이 'A'인 사람들의 평균 근속 연수를 계산하시오.
- 'B' : 부서가 'Sales'이고 등급이 'B'인 사람들의 평균 교육참가횟수를 계산하시오.
- A와 B를 더한 값을 구하시오.

```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_1_3.csv')
df.info()
```

```python
# @title
# 3-1. 결측치 처리
# 평균만족도 : 결측치는 평균만족도 컬럼의 전체 평균으로 채우시오.
# 근속연수 : 결측치는 각 부서와 등급별 평균 근속연수로 채우시오.(평균값의 소수점은 버림 처리)

print(df.isnull().sum())

df['평균만족도'] = df['평균만족도'].fillna(df['평균만족도'].mean())

df_merge = df.groupby(['부서', '등급'])['근속연수'].mean().astype(int).reset_index()
df = df.merge(df_merge, how='left', on=['부서', '등급'], suffixes =('', '_mean'))
df['근속연수'] = df['근속연수'].fillna(df['근속연수_mean'])

print(df.isnull().sum())
```

```python
# @title
# 3-2. 조건에 따른 평균 계산
# A : 부서가 'HR'이고 등급이 'A'인 사람들의 평균 근속 연수를 계산하시오.
# 'B' : 부서가 'Sales'이고 등급이 'B'인 사람들의 평균 교육참가횟수를 계산하시오.
# A와 B를 더한 값을 구하시오.

# 3-2 정답 : 25.225
A = df[(df['부서'] == 'HR') & (df['등급'] == 'A')]['근속연수'].mean()
B = df[(df['부서'] == 'Sales') & (df['등급'] == 'B')]['교육참가횟수'].mean()

print("A is", A)
print("B is", B)
print(A + B)
```

```python
import numpy as np
import pandas as pd

train = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_2_train.csv')
test = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_2_test.csv')

train.info()
```

```python
# @title
from sklearn.ensemble import RandomForestClassifier
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

test_ID = test['ID']

X_train = train.drop(['ID', '라벨'], axis = 1)
y_train = train['라벨']
X_test = test.drop(['ID', '라벨'], axis = 1)

cat_cols = X_train.select_dtypes('object').columns
num_cols = ['농업면적']

# print(OneHotEncoder.__doc__)

transformer = make_column_transformer(
    (OneHotEncoder(handle_unknown='ignore', sparse_output = False), cat_cols),
    (StandardScaler(), num_cols)
)

X_train_pre = transformer.fit_transform(X_train)
X_test_pre = transformer.transform(X_test)

rfc_model = RandomForestClassifier(random_state = 0)
rfc_model.fit(X_train_pre, y_train)

y_pred = rfc_model.predict(X_test_pre)

pd.DataFrame({'ID' : test_ID, 'pred' : y_pred}).to_csv('result.csv', index = False)
```

#### 🚩 1. 한 제조 회사에서 생산성을 높이고자 직원들의 주요 생산성 요인을 분석하기로 결정하였다. 이를 위해 200명의 직원 데이터를 수집했으며, 직원들의 근무기간, 특성정보, 그리고 개인적인 속성을 조사하였다.(9_3_1)

Data description
- id : 데이터의 고유 식별자
- tenure : 사용기간
- f2 : 고객의 두 번째 특성
- f3 : 고객의 세 번째 특성
- f4 : 고객의 네 번째 특성
- f5 : 고객의 다섯 번째 특성
- design : 생산성 점수

1-1. design을 예측하는 다중회귀 분석을 시행한 후 유의하지 않은 설명변수 개수를 구하시오. (단, 불필요한 칼럼은 제외하며, 모델의 절편항은 포함)

데이터 분리 조건
- 훈련 데이터 : id1 ~ id140
- 테스트 데이터 : id141 ~ id200

1-2. 훈련 데이터(학습용 데이터)의 예측값과 실제값의 피어슨 상관계수를 구하시오.(소수점 셋째 자리에서 반올림)

1-3. 적합한 모델을 활용하여 테스트 데이터에서의 RMSE를 구하시오.

```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_3_1.csv')
df.head()
```

```python
# @title
df['f3'].value_counts()
```

```python
# @title
from statsmodels.formula.api import ols

train = df[(df['id'] >= 1) & (df['id'] <= 140)]
test = df[~((df['id'] >= 1) & (df['id'] <= 140))]

model = ols('design ~ tenure + f2 + f3 + f4 + f5', data = train).fit()
# 유의하지 않은 설명변수 : 2개
# 여기에서 모델의 절편항은 포함하는 의미는 모델 구조를 짤 때 'design ~ tenure + f2 + C(f3) + f4 + f5 -1' 이런식으로 절편을 제거하지 말라는 뜻.
# model.pvalues > 0.05

pred_train = model.predict(train)
true_train = train['design']

from scipy.stats import pearsonr

p = pearsonr(true_train, pred_train)
# 1-2 정답 : 0.915
print(round(p.statistic, 3))

from sklearn.metrics import root_mean_squared_error

pred_test = model.predict(test)
true_test = test['design']

# 1-3 정답 : 4.396152958589427
print(root_mean_squared_error(true_test, pred_test))
```

```python
# @title
train = df[(df['id'] >= 1) & (df['id'] <= 140)]
test = df[~((df['id'] >= 1) & (df['id'] <= 140))]

train = train.drop('id', axis = 1)
test_x = test.drop(['id', 'design'], axis = 1)

from statsmodels.formula.api import ols

ols_model = ols('design ~ tenure + f2 + f3 + f4 + f5', data = train).fit()
# ols_model.summary()

# 1-1 정답 : 3개
# print(sum(ols_model.pvalues > 0.05))
# 유의하지 않은 설명변수 : 절편항, f3, f5

test_pred = ols_model.predict(test_x)

# ❗pearsonr 상관계수 함수 위치 잊지 말기!
from scipy.stats import pearsonr

p = pearsonr(test['design'], test_pred)
print(round(p.statistic, 3))
# 1-2 정답 : 0.943

from sklearn.metrics import root_mean_squared_error

rmse = root_mean_squared_error(test['design'], test_pred)
print(rmse)
# 1-3 정답 : 4.396152958589427
```

BMI = 몸무게 / 키^2 (소수점 첫째 자리까지 반올림)

| 구분     | BMI 기준          |
|----------|--------------------|
| 저체중   | 18.5 미만          |
| 정상     | 18.5 이상 23 미만  |

| 비만     | 25 이상            |

```python
import pandas as pd
import numpy as np

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/6_1_2.csv')
df.head(2)
```

```python
# @title
def convert(x):
    if x < 18.5:
        return "저체중"
    elif ((x >= 18.5) & (x < 23)):
        return "정상"
    elif ((x >= 23) & (x < 25)):
        return "과체중"
    elif x >= 25:
        return "비만"

df['BMI'] = round((df['Weight_kg'] / ((df['Height_cm'] * 0.01) ** 2)), 1)
df['정도'] = df['BMI'].apply(lambda x : convert(x))
print(len(df[(df['정도'] == '저체중') | (df['정도'] == '비만')]))
# 2. 74명
```

### 작업형 제 2유형

제공된 학습용 데이터(6_2_train)는 환자의 나이, 성별, 체질량지수(BMI), 혈당 및 콜레스테롤 수치 등의 정보를 포함하고 있으며, 이로부터 이완기 혈압(DBP)을 예측하고자 한다. 학습용 데이터를 활용하여 환자의 이완기 혈압(DBP)를 예측하는 회귀 모델을 개발하고, 가장 우수한 모델을 평가용 데이터(6_2_test)에 적용하여 예측 결과를 제출하시오.

% 모델 성능 지표 : Macro F1 Score
% level_0 칼럼은 인덱스 초기화 과정에서 생성된 것으로 분석 시 제외

Data description
- ID : 고유 식별자
- Age : 나이
- Gender : 성별(범주형 : Male/Female 등)
- BMI : 체질량지수
- ALT : 간 기능 수치
- FPG : 공복혈당
- Chol : 총 콜레스테롤
- Tri : 중성지방
- HDL : 고밀도지단백 콜레스테롤
- LDL : 저밀도지단백 콜레스테롤
- DBP : 이완기 혈압


제출형식
- 파일명 : result.csv (디렉토리/폴더명 제외)
- 제출칼럼 : ID, pred (총 2개 칼럼)
- pred : 예측된 이완기 혈압(DBP) 값(실수형 가능)
- 행 수 : 테스트 데이터(6_2_test)의 ID 수와 동일

```python
import pandas as pd
import numpy as np

train = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/6_2_train.csv')
test = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/6_2_test.csv')

train.head(2)
```

```python
# @title
import pandas as pd
import numpy as np
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.ensemble import RandomForestRegressor

train = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/6_2_train.csv')
test = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/6_2_test.csv')

# Gender 결측치 존재.

test_ID = test['ID']

mode_value = train['Gender'].mode()[0]

train['Gender'] = train['Gender'].fillna(mode_value)
test['Gender'] = test['Gender'].fillna(mode_value)

X_train = train.drop(['level_0', 'ID', 'DBP'], axis = 1)
y_train = train['DBP']

X_test = test.drop(['level_0', 'ID', 'DBP'], axis = 1)

cat_col = ['Gender']
num_col = X_train.select_dtypes('number').columns

transformer = make_column_transformer(
    (OneHotEncoder(handle_unknown='ignore', sparse_output = False), cat_col),
    (StandardScaler(), num_col),
    remainder = 'passthrough'
)

X_train_pre = transformer.fit_transform(X_train)
X_test_pre = transformer.transform(X_test)

model = RandomForestRegressor(random_state = 1)
model.fit(X_train_pre, y_train)

pred = model.predict(X_test_pre)

pd.DataFrame({'ID' : test_ID, 'pred' : pred}).to_csv('result.csv', index = False)
```

#### 🚩1. 어느 회사에서 100명의 직원들을 대상으로 하루 업무 수행 시간을 조사하였다. K-S 검정을 통해 업무 수행 시간이 정규분포를 따르는 지 검정하고자 한다.(6_3_1)

1-1. 직원들의 업무 수행 시간의 평균과 표준편차를 구하시오.(소수점 셋째 자리까지 반올림)

1-2. 직원들의 업무 수행 시간이 정규분포를 따르는 지 K-S 검정을 실시하고, 검정통계량을 계산하시오.(소수점 셋째 자리까지 반올림)

1-3. p-value를 바탕으로 유의수준 5%에서 귀무가설의 기각/채택 여부를 결정하시오.(p-value는 소수점 셋째 자리까지 반올림)

```python
import pandas as pd
import numpy as np

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/6_3_1.csv')
df.head(2)
```

```python
# @title
# 1-1.
mean = df['work_hours'].mean()
std = df['work_hours'].std()
print(round(mean, 3), round(std, 3))

# 1-2
from scipy.stats import kstest, norm
stat, pvalue = kstest(df['work_hours'], 'norm', args = (mean, std))
print("검정통계량", stat)

# 1-3
print("기각" if round(pvalue, 3) < 0.05 else "채택")
print(round(pvalue, 3))

# print(ks_1samp.__doc__)
```

#### 2. 다음의 데이터는 주택들의 가격(price), 면적(area), 방의 개수(rooms), 연식(age)을 조사하여 기록한 것이다.(6_3_2)

2-1. 주택 가격을 종속 변수로 하고, 면적, 방의 개수, 연식을 독립 변수로 하는 다중회귀 분석을 수행하여, 회귀 계수가 가장 높은 변수를 구하시오. (다중회귀모형 적합 시 절편 포함)

2-2. 유의수준 5% 하에서 각 독립 변수가 주택 가격에 미치는 영향이 통계적으로 유의미한 지 판단하고, 유의미한 변수 개수를 구하시오.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/6_3_2.csv')
df.head(2)
```

```python
# @title
from statsmodels.formula.api import ols

model = ols('price ~ area + rooms + age', data = df).fit()
# model.params
# 2-1. 회귀계수가 가장 높은 변수 : rooms

# 2-2. 유의한 변수 개수  : 3개. 전부 유의수준보다 작음.
model.pvalues < 0.05
```
